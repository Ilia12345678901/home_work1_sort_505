```python
from random import randint
s = 20
a = []
c = 0
for i in range (s) :
    c = randint(1,100)
    a.append(c)
print(a)
```

```python
def sel_sort (a : list) -> None :
    n = len(a)
    for i in range (n-1) :
        for j in range (i+1, n) :
            if a[i] > a[j] :
                a[i], a[j] = a[j], a[i]
sel_sort(a)
print(a)
```

```python
def bubble_sort_optimized(a : list) -> None :
    n = len(a)
    for i in range(n - 1, 0, -1):
        flag = False
        for j in range(i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                flag = True
        if not flag:
            break
bubble_sort_optimized(a)
print(a)
```

```python
def counting_sort(arr : list) -> None :
    max_val = max(arr)
    min_val = min(arr)
    l = [0]*(max_val - min_val + 1)
    for i in arr :
        l[i - min_val] += 1
    arr.clear()
    for j in range(min_val, max_val + 1) :
        ind = j - min_val
        for _ in range (l[ind]) :
            arr.append(j)
counting_sort(arr)
print(arr)
```
